#### Preprocess the interactions

```python
# only keep positive interactions
interactions = interactions[interactions['interaction_type'].isin([1,2,3,5])]
```

```python
# sort by time
interactions.sort_values(by='created_at', inplace=True)
```

```python
# convert timestamps to dates
interactions['date'] = pd.to_datetime(interactions['created_at'], unit='s')
max_date = interactions['date'].max() # find the latest timestamp
cutoff = max_date - pd.Timedelta(days=7) # define a cutoff of exactly one week
```

```python
before_cutoff = interactions[interactions["date"] <= cutoff]
after_cutoff = interactions[interactions["date"] > cutoff] # last week
```

```python
# now we need to filter the users such that they will have interactions both before and after the one week cutoff
# because we will take the target item from after the cutoff, but the history will be based on interactions before the cutoff
before_counts = before_cutoff.groupby('user_id').size()
after_counts = after_cutoff.groupby('user_id').size()

valid_users = before_counts[before_counts >= 3].index
valid_users = valid_users.intersection(after_counts[after_counts >= 1].index)

before_cutoff = before_cutoff[before_cutoff['user_id'].isin(valid_users)]
after_cutoff = after_cutoff[after_cutoff['user_id'].isin(valid_users)]
users = users[users['id'].isin(valid_users)]
```

```python
# check the total number of users now
len(users)
```

#### Construct the user history

```python
# build the user history by taking last 10 interactions before the cutoff
K = 10
history_dict = {}

for user in valid_users:
    user_hist = before_cutoff[before_cutoff['user_id'] == user]
    
    # already sorted ascending, take last K
    last_k_items = user_hist['item_id'].tolist()[-K:]
    
    history_dict[user] = last_k_items
```

```python
history_dict
```

```python
# build a dictionary with user_id, the user history, and the target item (which comes from the interactions of the last week)
rows = []

for _, row in after_cutoff.iterrows():
    user = row['user_id']
    target_item = row['item_id']
    
    if user not in history_dict:
        continue
        
    history = history_dict[user]
    
    rows.append({
        "user_id": user,
        "history": history,
        "target_id": target_item
    })
```

```python
# convert it to dataframe
final_df = pd.DataFrame(rows)
```

#### Convert history item ids to text

Here we need to convert the history (and also target_id) into informative text so that, in the future, when we embed the history, the embeddings will be informative enough for the LLM. Since we do not have a title for each job item, we can create one based on the industry and discipline, maybe also include the country. The issue is that the dataset uses anonymized IDs for the industry and discipline of each job posting. One solution here would be that we create synthetic mapping from the industry and discipline IDs to their name. There are 23 industry IDs, and 23 discipline IDs. 

#### Add synthetic gender and age

```python
# we want to keep the dataset balanced, so we add randomly 50% males and 50% females

np.random.seed(42)

unique_users = final_df['user_id'].unique()

gender_map = {user: np.random.choice(['Male', 'Female'])
              for user in unique_users}

final_df['gender'] = final_df['user_id'].map(gender_map)
```

```python
# check the distribution
final_df["gender"].value_counts(normalize=True)
```

Now, to add the synthetic age, we can take advantage that we know the total experience years of each user. This is indicative of the age range.

```python
# first add the experience years to final_df so we merge later easily
users = users.rename(mapper={'id':'user_id'}, axis='columns')
final_df = final_df.merge(
    users[['user_id','experience_years_experience']],
    on='user_id',
    how='left'
)
```

We have this mapping from the dataset documentation:  
  
0 = unknown.  
1 = less than 1 year  
2 = 1 - 3 years  
3 = 3 - 5 years  
4 = 5 - 10 years  
5 = 10 - 20 years  
6 = more than 20 years  
  
We remove the users with unknown experience years.   
  
We can assign the age range as follows:  
1 = 20 - 22 years old  
2 = 23 - 25 years old  
3 = 26 - 30 years old  
4 = 31 - 40 years old  
5 = 41 - 54 years old  
6 = 55+ years old  

```python
# first remove the users with unknow experience
final_df = final_df[final_df['experience_years_experience'] != 0]

# now create age map and age_range column
age_map = {1: '20-22', 2:'23-25', 3:'26-30', 4:'31-40', 5:'41-54', 6:'55+'}
final_df['age_range'] = final_df['experience_years_experience'].map(age_map)
```

```python
# remove the experience years since we do not need them anymore
final_df = final_df[['user_id', 'history', 'target_id', 'gender', 'age_range']]
```
